```python
#water container most area 

def water_max_area(heights):
    max_area = 0
    for left in range(len(heights)):
        for right in range(left + 1, len(heights)):
            width = right - left
            height = min(heights[left], heights[right])
            area = height * width 
            if area > max_area:
                max_area = area 
    return max_area

h = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print(water_max_area(h))
```

```python
#water trap in between

def trap_water(heights):
    left_max = 0
    right_max = 0
    water = 0

    #find left and right max for each index
    for i in range(len(heights)):
        for j in range(i):
            if left_max < heights[j]:
                left_max = heights[j]
        for x in range(i + 1, len(heights), 1):
            if right_max < heights[x]:
               right_max = heights[x]
    #water fill levels
        water_val = min(left_max, right_max) - heights[i]
        if water_val > 0:
            water += water_val

    return water

h = [4, 2, 0, 3, 2, 5]
print(trap_water(h)) 
```
