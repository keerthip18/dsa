```python
#find the midpoint

def midpoint(list):
    length = len(list)
    return list[length // 2]

print(f"odd: {midpoint(list_odd)}")
print(f"even upper: {midpoint(list_even)}")
```

```python
#find boundaries of nums

def boundaries(list):
    left = 0 
    right = len(list) - 1
    return left, right

print(boundaries(list_odd))
print(boundaries(list_even))
```

```python
#find midpoint using boundaries

def midpoint_boundaries(list):
    if list:
        left, right = boundaries(list)
        midpoint = (right - left) // 2
        if (len(list) % 2 == 0):
            even_lower = list[midpoint]
            even_upper = list[midpoint + 1]
            return (even_lower, even_upper)
        return (list[midpoint])
    return

list_odd = []
list_even = [10, 20]
print(f"odd: {midpoint_boundaries(list_odd)}")
print(f"even lower, even upper: {midpoint_boundaries(list_even)}")
```

```python
#palidrome

def is_palindrome(word):
    word_list = list(word)
    left = 0
    right = len(word_list) - 1
    while left < right:
        if word_list[left] != word_list[right]:
            return False
        left += 1
        right -= 1
    return True 

print(is_palindrome("free"))
```

```python
#reverse nums

def reverse(list):
    left = 0
    right = len(list) - 1
    while left < right:
        list[left], list[right] = list[right], list[left]
        left += 1
        right -= 1
    return list
l = [10, 20]
print(reverse(l))
```

```python
#two sum with sorted list

def two_sum_sorted(list, target):
    left = 0
    right = len(list) - 1
    while left < right:
        val = list[left] + list[right]
        if val < target:
            left = left + 1 
        elif val > target:
            right = right - 1
        else:
            return list[left], list[right]
l = [10]
print(two_sum_sorted(l, 30)) 
```

```python
#two sum with non-sorted list

def two_sum_not_sorted(nums, target):
    seen = []
    for num in nums:
        complement = target - num
        if complement in seen:
            return complement, num
        seen.append(num)
    return None 

l = [2, 11, 7, 15]
print(two_sum_not_sorted(l, 9)) 
```

```python
#binary search

def binary_search(list, target):
    left = 0
    right = len(list) - 1
    while left < right:
        midpoint = left + (right - left) // 2
        if list[midpoint] > target:
            right = midpoint - 1 
        elif list[midpoint] < target:
            left = midpoint + 1
        else:
            return midpoint
    return -1

l = [2, 7, 11, 15]
print(binary_search(l, 11))
```
