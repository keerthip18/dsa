```python
#remove duplicates in sorted array
def remove_duplicates(list):
    slow = 0
    for fast in range(1, len(list)):
        if list[slow] != list[fast]:
            slow = slow + 1
            list[slow] = list[fast]
    return list[:slow + 1]

l = [1, 1, 2, 2, 3, 4, 4, 5, 6]
print(remove_duplicates(l))
```

```python
#remove target in array
def remove_target(list, target):
    slow = 0
    for fast in range(len(list)):
        if list[fast] != target:
            list[slow] = list[fast]
            slow += 1
    return list[:slow]

l = [1, 1, 2, 2, 3, 4, 4, 5, 6]
print(remove_target(l, 2))
```

```python
#move zeros to end of array
def move_zeros(list):
    slow = 0
    for fast in range(len(list)):
        if list[fast] != 0:
            list[slow], list[fast] = list[fast], list[slow]
            slow += 1
    return list

l = [2, 3, 1, 1, 0, 2, 0, 3]
print(move_zeros(l))
```
