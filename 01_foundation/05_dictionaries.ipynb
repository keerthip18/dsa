```python
# 1. Creation: Defining a dictionary with {key: value}
user_data = {
    "name": "Alice",
    "age": 25,
    "city": "Austin"
}
print("Initial dictionary:", user_data)
```

```python
# 2. Accessing: Getting a value using its key inside square brackets
user_name = user_data["name"]
print("Name is:", user_name)
```

```python
# 3. Adding/Updating: Assigning a value to a key
# If the key exists, it updates; if not, it creates it.
user_data["job"] = "Engineer"  # Adding
user_data["age"] = 26          # Updating
print("After adding job and updating age:", user_data)
```

```python
# 4. Membership: Checking if a KEY exists using 'in'
has_city = "city" in user_data
print("Does 'city' exist?:", has_city)
```

```python
# 5. Deletion: Removing a key-value pair using 'del'
del user_data["city"]
print("After deleting 'city':", user_data)
```

```python
# 6. Length: Checking how many pairs are inside
size = len(user_data)
print("Number of items:", size)
```

```python
# 7. Merging: Joining two dictionaries using the '|' operator (Python 3.9+)
extra_info = {"hobby": "Hiking", "pet": "Cat"}
full_profile = user_data | extra_info
print("Merged dictionary:", full_profile)
```
