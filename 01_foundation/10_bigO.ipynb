```python
def time_logN(iterations):
    counter = 0
    while iterations > 1:
       counter += 1
       iterations = iterations // 2
    return counter

num = time_logN(1_000_000_000_000)
print(num)
#1000 (1 thousand) = 10 operations
#1_000_000 (1 million) = 20 operations
#1_000_000_000 (1 billion) = 30 operations
#1_000_000_000_000 (1 trillion) = 40 operations
```

```python
def time_n2(iterations):
    counter = 0
    for i in range(iterations):
        for j in range(iterations):
            counter += 1
    return counter

num = time_n2(2)
print(num)
```

```python
def time_n3(iterations):
    counter = 0
    for i in range(iterations):
        for j in range(iterations):
            for x in range(iterations):
                counter += 1
    return counter

num = time_n3(1000)
print(num)
```

```python
def time_nlogN(iterations):
    counter = 0
    for i in range(iterations):
        iter = iterations
        while iter > 1:
            counter += 1
            iter = iter // 2
    return counter 

num = time_nlogN(1_000_000)
print(num) 
```

```python
counter = 1
def time_2PowerN(iterations):
    global counter 
    if iterations == 0:
        return 
    counter += 1
    time_2PowerN(iterations - 1) 
    time_2PowerN(iterations - 1)

time_2PowerN(20)
print(counter)
```

```python
counter = 0
def time_nFactorial(iterations):
    global counter
    if iterations == 0:
        counter += 1  
        return
    for i in range(iterations):
        time_nFactorial(iterations - 1)  

time_nFactorial(4)
print(counter) 
```
