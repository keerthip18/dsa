```python
# 1. Creation: Using parentheses (or just commas)
my_tuple = (10, 20, 30, 40, 50)
single_item_tuple = (5,) # Note the comma! Without it, Python thinks it's just an integer.
print("Initial tuple:", my_tuple)
```

```python
# 2. Indexing: Accessing values by position
first = my_tuple[0]
last = my_tuple[-1]
print("First and Last:", first, last)
```

```python
# 3. Slicing: Extracting a portion
middle_part = my_tuple[1:4]
print("Slice of tuple:", middle_part)
```

```python
# 4. Concatenation: Joining two tuples to create a NEW one
# Remember: This doesn't change 'my_tuple', it makes a third tuple.
combined = my_tuple + (60, 70)
print("Combined tuple:", combined)
```

```python
# 5. Repetition: Multiplying the tuple
repeated = ("A", "B") * 3
print("Repeated tuple:", repeated)
```

```python
# 6. Membership: Checking if an item exists
is_present = 30 in my_tuple
print("Is 30 in the tuple?:", is_present)
```

```python
# 7. Unpacking: Spreading values into variables
x, y, z, *rest = my_tuple
print("Unpacked x, y, z:", x, y, z)
print("The 'rest' packed into a list:", rest)
```

```python
# 8. Length: Counting items
count = len(my_tuple)
print("Length of tuple:", count)
```
