```python
# 1️⃣ Creating strings
s1 = "Hello"
s2 = "World"
empty = ""
print("Strings:", s1, s2, empty)
```

```python
# 2️⃣ Accessing characters (indexing)
first_char = s1[0]
last_char = s1[-1]
print("First and last characters:", first_char, last_char)
```

```python
# 3️⃣ Modifying strings
# Strings are immutable! You cannot directly change a character
# But you can create a new string using slices + concatenation
s3 = s1[:2] + "y" + s1[3:]
print("Modified string:", s3)  # 'Heylo'
```

```python
# 4️⃣ Concatenation using +
combined = s1 + " " + s2
print("Concatenated string:", combined)
```

```python
# 5️⃣ Repeating strings using *
repeat = s1 * 3
print("Repeated string:", repeat)
```

```python
# 6️⃣ Slicing strings
sub1 = s1[1:4]      # 'ell'
sub2 = s1[:3]       # 'Hel'
sub3 = s1[2:]       # 'llo'
sub4 = s1[::2]      # 'Hlo'
reverse_s1 = s1[::-1] # 'olleH'
print("Sliced strings:", sub1, sub2, sub3, sub4, reverse_s1)
```

```python
# 7️⃣ Membership testing
print("'H' in s1?", 'H' in s1)
print("'z' in s1?", 'z' in s1)
```

```python
# 8️⃣ Combining operations
new_string = (s1 + " " + s2)[::2] * 2
print("Combined operations:", new_string)
```

```python
# 9️⃣ Replacing part of a string using slicing
# Strings are immutable, so we create a new string
s4 = s1[:1] + "a" + s1[2:]
print("Slice-replaced string:", s4)  # 'Hallo'
```

```python
# 1️⃣0️⃣ Copying a string
s_copy = s1[:]
print("Copied string:", s_copy)
```
