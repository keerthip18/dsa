```python
list = [1, 2, 3, 4, 5]
print(f"{list}, address = {id(list)}, hex = {hex(id(list))}")
for i, num in enumerate(list):
    print(f"index {i}: {num}, address = {hex(id(num))}")
```

```python
a = 10
b = -5
print(a, type(a))  # 10 <class 'int'>
print(b, type(b))  # -5 <class 'int'>
```

```python
c = 3.14
d = -0.5
print(c, type(c))  # 3.14 <class 'float'>
print(d, type(d))  # -0.5 <class 'float'>
```

```python
z = 2 + 3j
print(z, type(z))  # (2+3j) <class 'complex'>
```

```python
name = "Python"
print(name, type(name))
print(name[0])     # P (first character)
print(name[-1])    # n (last character)
```

```python
fruits = ["apple", "banana", "cherry"]
print(fruits, type(fruits))
fruits.append("orange")  # add item
print(fruits)
```

```python
point = (10, 20)
print(point, type(point))
# point[0] = 5  -> ERROR! tuples cannot be changed
```

```python
numbers = {1, 2, 3, 2, 1}
print(numbers)  # {1, 2, 3}  (duplicates removed)

numbers.add(4)
print(numbers)
```

```python
student = {"name": "Alice", "age": 20}
print(student, type(student))
print(student["name"])  # Alice
student["age"] = 21     # update value
print(student)
```
