```python
# 1. Creation: Using curly braces {} or set() for an empty set
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print("Set A:", set_a)
print("Set B:", set_b)
```

```python
# 2. Membership: Checking if an item exists (Extremely fast in sets)
print("Is 1 in Set A?:", 1 in set_a)
```

```python
# 3. Union (|): Combining everything from both sets (no duplicates)
# This is like 'OR'
all_items = set_a | set_b
print("Union (A or B):", all_items)
```

```python
# 4. Intersection (&): Finding only what they have in common
# This is like 'AND'
common_items = set_a & set_b
print("Intersection (A and B):", common_items)
```

```python
# 5. Difference (-): Items in A that are NOT in B
only_in_a = set_a - set_b
print("Difference (Only in A):", only_in_a)
```

```python
# 6. Symmetric Difference (^): Items in EITHER A or B, but NOT both
# It removes the overlap
unique_to_each = set_a ^ set_b
print("Symmetric Difference:", unique_to_each)
```

```python
# 7. Comparison Operators: Checking subsets
small_set = {1, 2}
is_subset = small_set <= set_a  # True if all items of small_set are in set_a
print("Is {1,2} a subset of A?:", is_subset)
```
