```python
# 1️⃣ Creating lists (literal)
a = []          # empty list
b = [1, 2, 3]   # list with elements
print("Initial lists:", a, b)
```

```python
# 2️⃣ Accessing elements (indexing)
first = b[0]
last = b[-1]
print("First element:", first)
print("Last element:", last)
```

```python
# 3️⃣ Modifying elements (assignment)
b[1] = 99
print("After modification:", b)
```

```python
# 4️⃣ Adding elements using concatenation
b = b + [4]      # add single element
b = b + [5, 6]   # add multiple elements
print("After concatenation:", b)
```

```python
# 5️⃣ Repeating lists using *
c = [7] * 3
print("Repeated list:", c)
```

```python
# 6️⃣ Removing elements using slicing (no remove functions)
# Remove index 2
b = b[:2] + b[3:]
print("After removing index 2:", b)
```

```python
# 7️⃣ Slicing (sub-lists)
sublist1 = b[1:4]    # elements from index 1 to 3
sublist2 = b[:3]     # first three elements
sublist3 = b[2:]     # elements from index 2 to end
sublist4 = b[::2]    # every second element
reverse_b = b[::-1]  # reversed list
print("Slicing examples:", sublist1, sublist2, sublist3, sublist4, reverse_b)
```

```python
# 8️⃣ Checking membership (in / not in)
print("Is 99 in b?", 99 in b)
print("Is 100 in b?", 100 in b)
```

```python
# 9️⃣ Assigning to slices (inserting, replacing, deleting)
# Replace elements at index 1 and 2
b[1:3] = [8, 8]
print("After slice replacement:", b)

# Insert elements at index 2
b[2:2] = [20, 21]
print("After slice insertion:", b)

# Delete elements at index 0 to 1
b[0:2] = []
print("After slice deletion:", b)
```

```python
# 10️⃣ Copying a list using slicing
b_copy = b[:] 
print("Copied list:", b_copy)
```

```python
# 11️⃣ Combining all operators
combined = b + c * 2 + [99]
print("Combined operation result:", combined)
```

```python
a = [1, 2, 3]
b = a[1:2]
b[0] = 5
b = b + [4, 4]
print("a list:", a)
print("b list:", b)
```

```python
# Unpacking: Assigning items to variables
my_list = [10, 20, 30, 40]
a, b, c, d = my_list # Note: Number of variables must match list length
print("Unpacked values:", a, b, c, d)
```

```python
a, b = my_list[0:2]
print("Unpacked values:", a, b)
```

```python
a, b, *c = my_list
print("Unpacked values:", a, b, c)
```

```python
a = [1, 2, 3]
b = [3, 4, 5]
new_list = a + b

print("combined list:", new_list)
```
