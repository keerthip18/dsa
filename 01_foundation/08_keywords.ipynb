```python
# --- 1. Logic & Booleans ---
# True, False, None, and, or, not, is, in
a = True
b = False
c = None  # Represents 'nothing'

if a and not b:  # 'and', 'not'
    print("Logic checks out!")

if c is None:    # 'is' checks exact identity
    print("c is empty")
```

```python
# --- 2. Structural/Control Flow ---
# if, elif, else, for, while, break, continue, pass
numbers = [1, 2, 3]
if 5 in numbers: # 'in' checks membership
    pass         # 'pass' is a do-nothing placeholder
elif 1 in numbers:
    for n in numbers:
        if n == 2:
            continue # Skip the rest of this loop iteration
        print("Looping:", n)
else:
    
    print("No match")
```

```python
# --- 3. Functions & Scopes ---
# def, return, lambda, global, non-local
def my_function(x): # 'def' defines a function
    return x * 2    # 'return' sends a value back

sq = lambda x: x * x # 'lambda' is an anonymous one-line function
```

```python
# --- 4. Exception Handling ---
# try, except, finally, raise, assert
try:
    # assert 1 == 2, "Math broke!" # 'assert' tests a condition
    result = 10 / 0
except ZeroDivisionError: # 'except' catches the error
    print("Can't divide by zero")
finally:
    print("This always runs") # 'finally' runs no matter what
```

```python
# --- 5. Data & Objects ---
# del, class, import, from, as, with
import math as m     # 'import' and 'as' (aliasing)
from math import pi  # 'from' imports specific parts

del numbers          # 'del' removes the reference to the list
```
