# List Comprehensions

A list comprehension builds a new list in one line:

```
[expression for item in iterable if condition]
```

````python
# Transforming strings

words = ["hello", "world", "python"]
capitalized = [w.upper() for w in words]
print(words)
print(capitalized)
````

````python
a = range(1, 11)
output = ["even" if num % 2 == 0 else "odd" for num in a]
print(output)
````

````python
email = "bob@yahoo.com"
email.split("@")[1]
````

````python
emails = [
    "alice@gmail.com",
    "bob@yahoo.com",
    "charlie@gmail.com",
    "david@outlook.com"
]

output = [email.split("@")[1] for email in emails]
print(output)
````

````python
a = range(1,11)
printA = [print(num) for num in a]
even_squares = [num * num for num in a if num % 2 == 0]
print(even_squares)
````

````python
a = [1, 2, 3]
sq = [num * num for num in a]
print(sq)
````

````python
def cap(word):
    word = word[0].upper() + word[1:]
    return word

lambdaCap = lambda word: word[0].upper() + word[1:]
````

````python
words = ["hello", "world", "python"]
capitalized = [lambdaCap(w) for w in words if(len(w)) <= 5]
print(capitalized)
````

````python
word = "hello"
capitalizedWord = word[0].upper() + word[1:]
print(capitalizedWord)
````

````python
words = ["hello", "world", "python"]
i = 0
for w in words:
    words[i] = w.upper()
    i = i + 1
print(words)
````

````python
words = ["hello", "world", "python"]
capitalized = []
for w in words:
    capitalized.append(w.upper())
print(capitalized)
````

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix)
 
evenList = [num for row in matrix for num in row if num % 2 == 0]
print(evenList)
````

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flattenedMatrix = []
for row in matrix:
    for num in row:
       flattenedMatrix.append(num)
print(flattenedMatrix)

flattenedMatrix = [num for row in matrix for num in row]
print(flattenedMatrix)
````
