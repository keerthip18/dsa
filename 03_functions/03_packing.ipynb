# Packing and unpacking

- **Packing**: collecting several values into one variable (a tuple, list or dict).
- **Unpacking**: spreading one collection out into several variables or arguments.

The `*` operator works with sequences (lists, tuples, strings…), and `**` works with dicts.

## 1. Tuple packing and unpacking

```python
a, b = 1, 2
print(a, b)
```

```python
x, y, z = 3, 4, "Keerthi"
print(x, y, z)
```

```python
t = 1, 2, 3, 4, 5, 6
v1, v2, *v = t
print(v1, v2)
```

```python
# packing: comma-separated values become a tuple automatically
point = 3, 4
print(point, type(point))   # (3, 4) <class 'tuple'>

# unpacking: one variable per item
x, y = point
print(x, y)                 # 3 4
```

```python
# unpacking works with any iterable: lists, strings, ranges...
a, b, c = [1, 2, 3]
print(a, b, c)

first, second = "hi"
print(first, second)        # h i
```

```python
# the number of variables must match the number of items
try:
    a, b = [1, 2, 3]
except ValueError as e:
    print("ValueError:", e)
```

```python
# swapping: pack the right side into a tuple, then unpack into the left side
a, b = 1, 2
a, b = b, a
print(a, b)   # 2 1
```

## 2. Extended unpacking with `*`

A `*name` variable collects all the leftover items into a **list**.

```python
nums = [1, 2, 3, 4, 5]

first, *rest = nums
print(first, rest)          # 1 [2, 3, 4, 5]

*start, last = nums
print(start, last)          # [1, 2, 3, 4] 5

head, *middle, tail = nums
print(head, middle, tail)   # 1 [2, 3, 4] 5

a, b, *others = [1, 2]
print(others)               # [] - can be empty
```

```python
# _ is the usual name for values you don't care about
name, _, city = ("John", 25, "Dallas")
print(name, city)

first, *_ = nums
print(first)
```

## 3. Unpacking in loops

```python
# you've already used this with enumerate() and dict.items()
pairs = [("a", 1), ("b", 2), ("c", 3)]
for letter, number in pairs:
    print(letter, number)

person = {"name": "John", "age": 25}
for key, value in person.items():
    print(f"{key} → {value}")
```

```python
# nested unpacking: match the shape of the data
data = [("John", (25, "Dallas")), ("Amy", (30, "Austin"))]
for name, (age, city) in data:
    print(f"{name} is {age} and lives in {city}")
```

## 4. Packing function arguments: `*args` and `**kwargs`

In a function **definition**, `*` and `**` *pack* the extra arguments.

```python
# *args packs extra positional arguments into a tuple
def total(*args):
    print(args, type(args))
    return sum(args)

print(total(1, 2, 3))     # (1, 2, 3) <class 'tuple'>  then 6
print(total())            # () then 0
```

```python
# **kwargs packs extra keyword arguments into a dict
def describe(**kwargs):
    print(kwargs, type(kwargs))
    for key, value in kwargs.items():
        print(f"{key} → {value}")

describe(name="John", age=25, city="Dallas")
```

```python
# all together: normal parameters first, then *args, then **kwargs
def show(greeting, *args, **kwargs):
    print("greeting:", greeting)
    print("args:    ", args)
    print("kwargs:  ", kwargs)

show("hi", 1, 2, 3, color="red", size=10)
```

## 5. Unpacking function arguments: `*` and `**` in a call

In a function **call**, `*` and `**` do the opposite: they *unpack* a collection into separate arguments.

```python
def add(a, b, c):
    return a + b + c

nums = [1, 2, 3]
print(add(*nums))          # same as add(1, 2, 3)

values = {"a": 10, "b": 20, "c": 30}
print(add(**values))       # same as add(a=10, b=20, c=30)
```

```python
# print with * prints the items separated by spaces instead of printing the list
nums = [1, 2, 3]
print(nums)       # [1, 2, 3]
print(*nums)      # 1 2 3
print(*nums, sep=" | ")
```

## 6. Unpacking into new lists and dicts

```python
# merge lists
a = [1, 2]
b = [3, 4]
print([*a, *b])          # [1, 2, 3, 4]
print([0, *a, 99, *b])   # [0, 1, 2, 99, 3, 4]

# turn any iterable into a list or set
print([*"hello"])        # ['h', 'e', 'l', 'l', 'o']
print({*"hello"})        # set of unique letters
```

```python
# merge dicts - later keys win
defaults = {"color": "blue", "size": 10}
custom = {"size": 12}
settings = {**defaults, **custom}
print(settings)          # {'color': 'blue', 'size': 12}
```

## 7. `zip()`: pairing items from several iterables

`zip()` takes two or more iterables and pairs up their items **by position**: first with first, second with second, and so on. Each pair comes out as a **tuple**.

```python
names = ["John", "Amy", "Raj"]
ages = [25, 30, 28]

pairs = zip(names, ages)
print(pairs)          # <zip object ...> - an iterator, like map
print(list(pairs))    # [('John', 25), ('Amy', 30), ('Raj', 28)]
print(list(pairs))    # [] - used up, like every iterator
```

```python
# most common use: loop over two lists together, unpacking each tuple
for name, age in zip(names, ages):
    print(f"{name} is {age}")
```

```python
# what zip does by hand - compare with the loop above
for i in range(len(names)):
    print(f"{names[i]} is {ages[i]}")
```

```python
# three (or more) iterables -> tuples of three
cities = ["Dallas", "Austin", "Houston"]
for name, age, city in zip(names, ages, cities):
    print(f"{name:<5} {age}  {city}")
```

```python
# different lengths: zip stops at the shortest
print(list(zip([1, 2, 3, 4], ["a", "b"])))   # [(1, 'a'), (2, 'b')] - 3 and 4 are dropped

# strict=True (Python 3.10+) raises an error instead of silently dropping items
try:
    list(zip([1, 2, 3, 4], ["a", "b"], strict=True))
except ValueError as e:
    print("ValueError:", e)
```

```python
# zip_longest keeps going and fills the gaps
from itertools import zip_longest
print(list(zip_longest([1, 2, 3, 4], ["a", "b"], fillvalue="-")))
```

```python
# build a dict from a list of keys and a list of values
keys = ["name", "age", "city"]
values = ["John", 25, "Dallas"]
person = dict(zip(keys, values))
print(person)   # {'name': 'John', 'age': 25, 'city': 'Dallas'}
```

```python
# with enumerate: index + items from two lists
for i, (name, age) in enumerate(zip(names, ages), start=1):
    print(i, name, age)
```

```python
# unzip: zip(*pairs) splits a list of pairs back into separate tuples
pairs = [("John", 25), ("Amy", 30), ("Raj", 28)]
names2, ages2 = zip(*pairs)
print(names2)   # ('John', 'Amy', 'Raj')
print(ages2)    # (25, 30, 28)

# why it works: *pairs unpacks the list, so this is the same call
print(list(zip(("John", 25), ("Amy", 30), ("Raj", 28))))
```

```python
# zip(*matrix) turns rows into columns (transpose)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
]
for column in zip(*matrix):
    print(column)   # (1, 4)  (2, 5)  (3, 6)
```

```python
# pair each item with the next one
nums = [10, 15, 30, 32]
for a, b in zip(nums, nums[1:]):
    print(f"{a} -> {b}: change {b - a}")
```

## Summary

| Where | `*` | `**` |
|---|---|---|
| Left side of `=` | collect leftovers into a list | — |
| Function definition | pack extra positional args into a tuple | pack extra keyword args into a dict |
| Function call | unpack a sequence into positional args | unpack a dict into keyword args |
| Inside `[ ]`, `( )`, `{ }` | spread items into a new list/tuple/set | spread key-value pairs into a new dict |

`zip(a, b)` pairs items by position; `zip(*pairs)` unzips them again.
