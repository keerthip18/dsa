# Iterators

- An **iterable** is anything you can loop over: a list, string, dict, tuple, set or range.
- An **iterator** is an object that hands out items **one at a time** and remembers where it is.

You turn an iterable into an iterator with `iter()`, and get the next item with `next()`.

```python
# iter() and next()
nums = [10, 20, 30]
it = iter(nums)     # list (iterable) -> iterator

while True:
    n = next(it, None)
    if n is None:
        break
    print(n)

```

```python
# when the iterator runs out, next() raises StopIteration
try:
    print(next(it))
except StopIteration:
    print("no more items")

# next() with a default value returns the default instead of raising
print(next(it, "done"))
```

## What a `for` loop really does

A `for` loop is just `iter()` + `next()` + catching `StopIteration`. These two cells do the same thing.

```python
# the normal way
for n in nums:
    print(n)
```

```python
# what Python does behind the scenes
it = iter(nums)
while True:
    try:
        n = next(it)
    except StopIteration:
        break
    print(n)
```

## Iterators get used up

A list can be looped over again and again, because each `for` loop calls `iter()` and gets a fresh iterator. An iterator can only be gone through **once**.

```python
it = iter(nums)
print(list(it))   # [10, 20, 30]
print(list(it))   # []  - already used up

# a list makes a fresh iterator every time, so it is never used up
print(list(nums))
print(list(nums))
```

```python
def map1(func, lists):
   return (func(num) for num in lists)
```

```python
list1 = [1, 2]
```

```python
list2 = map1(lambda x: x * x, list1)
print(list(list2))
```

```python
list2 = [num * num for num in list1]
print(list2)
```

```python
def map2(x):
    return x + 2
```

```python
# map and filter return iterators too, so they also get used up
squares = map1(lambda x: x * x, nums)
print(squares)          # <map object ...>  - lazy, nothing computed yet
print(list(squares))    # [100, 400, 900]
print(list(squares))    # []
```

```python
import math
list1 = [10, 100, 1000]
logs = map(lambda x: math.log10(x), list1)
print(list(logs))
```

```python
squares = map(lambda x: x * x, range(1, 11, 1))
print(list(squares))
```

```python
squares = map(lambda x: x * x, range(1, 11, 1))
evenSquares = filter(lambda x: x % 2 == 0, squares)
print(list(evenSquares))
```

```python
nums = range(1, 11, 1)
evenSquares = [num * num for num in nums if num % 2 == 0]
print(evenSquares)
```

## Making your own iterator (class)

An iterator is any object with two methods:
- `__iter__()` returns the iterator itself
- `__next__()` returns the next item, or raises `StopIteration` when finished

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

for n in Countdown(5):
    print(n)
```

## The easy way: generators

A function with `yield` is a **generator**. Python writes `__iter__` and `__next__` for you. Each `yield` hands out one value and pauses the function until the next `next()` call.

```python
def countdown(start):
    while start > 0:
        yield start
        start -= 1

gen = countdown(3)
print(next(gen))   # 3
print(next(gen))   # 2
print(next(gen))   # 1

for n in countdown(5):
    print(n)
```

```python
# generator expression: like a list comprehension but with ( ), and lazy
squares = (x * x for x in range(1, 6))
print(squares)        # <generator object ...>
print(sum(squares))   # 55 - values are made one at a time, never stored in a list
```

## Why iterators matter

They are **lazy**: they make one item at a time instead of building everything up front. That saves memory and even lets you have endless sequences.

```python
def naturals():
    n = 1
    while True:      # never ends - fine, because values are only made when asked for
        yield n
        n += 1

for n in naturals():
    if n > 5:
        break
    print(n)
```
