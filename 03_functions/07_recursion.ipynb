```python
cntr = 0
cache = {}
def fib(n):
    global cntr 
    cntr += 1 
    # base cases: fib(0) = 0, fib(1) = 1
    if n <= 1:
        return n
    # recursive case: sum of the two previous numbers
    if(n in cache):
        return cache[n]
    else: 
        fib_num = fib(n - 1) + fib(n - 2)
        cache[n] = fib_num
        return cache[n]

num = fib(6)
print(f"fib : {num}")
print(f"counter: {cntr}")
```

```python
def fib_loop(n):
    if n <= 1:
        return n

    prev, curr = 0, 1              
    for _ in range(2, n + 1):
        prev, curr = curr, prev + curr  
    return curr

num = fib_loop(6)
print(f"fib : {num}")   
```
