In functional programming, functions are **values**. You can store them in variables, pass them to other functions, and return them from functions. A function that takes or returns another function is called a **higher-order function**.

```python
# apply takes a function and a value, and calls the function on the value
def apply(func, value):
    return func(value)
```

```python
# pass a named function (no parentheses: we pass the function itself, not its result)
def square(x):
    return x * x

def double(x):
    return x * 2

x = apply(square, 5)   # 25
y = apply(double, 5)  # 10
print(f"square : {x}")
print(f"double : {y}")
```

```python
# pass a lambda (a small anonymous function)
f1 = lambda x: x + 100
f2 = lambda s: s[::-1]
num = apply(f1, 5)     # 105
string = apply(f2, "hello") # olleh
print(num)
print(string)
```

```python
# pass built-in functions
print(apply(len, "Python"))     # 6
print(apply(str.upper, "hi"))   # HI
print(apply(abs, -42))          # 42
```

```python
# functions are values: store them in a list and loop over them
funcs = [square, double, abs, str]
for f in funcs:
    print(f"{f.__name__:<7} → {apply(f, -3)!r}")
```
