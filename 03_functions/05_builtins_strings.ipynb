# Built-in functions for strings

There are two kinds of tools for working with strings:

- **Built-in functions** like `len(s)` and `sorted(s)`, which you call with the string as an argument. Most also work with lists and other sequences.
- **String methods** like `s.upper()`, which belong to the `str` type and are called with a dot. There are 47 of them.

This notebook covers both.

**Strings are immutable.** No function or method changes a string in place. They all return a **new** string, so you have to keep the result: `s = s.upper()`.

# Part A: Built-in functions

## `str()`, `repr()`, `ascii()`: turning values into text

```python
print(str(42) + "!")          # numbers must be converted before joining with +
print(str(3.5), str(True), str(None), str([1, 2]))

# str() is the friendly version; repr() is the exact, developer version
text = "hi\tthere"
print(str(text))     # hi	there   (tab shown as a gap)
print(repr(text))    # 'hi\tthere' (quotes, and the tab shown as \t)

# ascii() is like repr() but escapes any non-English characters
print(ascii("café €"))   # 'caf\xe9 \u20ac'
```

```python
# !r in an f-string calls repr() - great for spotting hidden spaces
name = "John  "
print(f"[{name}]")
print(f"{name!r}")
```

## `len()`

```python
print(len("hello"))     # 5
print(len(""))          # 0
print(len("hi there"))  # 8 - spaces count
print(len("a\nb"))      # 3 - \n is one character
print(len("café"))      # 4 - counts characters, not bytes
print(len("café".encode()))  # 5 - é takes 2 bytes in UTF-8
```

## `ord()` and `chr()`

```python
print(ord("A"), chr(65))     # 65 A
print([ord(c) for c in "Hi!"])
print("".join(chr(n) for n in [80, 121, 116, 104, 111, 110]))
```

## `format()`

```python
# format(value, spec) - the same mini-language as f-strings
print(format("hi", "<10") + "|")    # left-align in 10 spaces
print(format("hi", ">10") + "|")    # right-align
print(format("hi", "^10") + "|")    # center
print(format("hi", "*^10"))         # center, fill with *
print(format(1234567.891, ",.2f"))  # 1,234,567.89
print(format(0.256, ".1%"))         # 25.6%
```

## `print()` and `input()`

```python
print("a", "b", "c")                 # separated by spaces
print("a", "b", "c", sep="-")        # custom separator
print("no newline", end=" | ")       # custom line ending
print("same line")
print(*"abc", sep=",")               # unpack the string's characters

# input() always returns a STRING, even if the user types a number:
# age = input("Age? ")       # user types 25  -> "25"
# age = int(input("Age? "))  # convert it yourself
```

## Sequence functions that work on strings

A string is a sequence of characters, so these all work on it. Note that most return a list or iterator, not a string.

```python
word = "python"

print(sorted(word))                 # ['h', 'n', 'o', 'p', 't', 'y'] - a LIST
print("".join(sorted(word)))        # hnopty - join it back into a string
print(list(reversed(word)))         # iterator of characters
print("".join(reversed(word)), word[::-1])   # two ways to reverse
print(min(word), max(word))         # by character code: h y
print(list(enumerate("abc")))
print(list(zip("abc", "123")))
```

```python
# any / all with a generator: check every character
password = "Secret123"
print(any(c.isdigit() for c in password))   # has a digit?
print(any(c.isupper() for c in password))   # has an uppercase letter?
print(all(c.isalnum() for c in password))   # letters and digits only?

# map / filter
print(list(map(str.upper, ["a", "b"])))
print("".join(filter(str.isdigit, "phone: 555-0142")))   # 5550142
```

```python
# comparing strings: compared character by character, by character code
print("apple" < "banana")   # True
print("Zebra" < "apple")    # True! uppercase letters come before lowercase
print("10" < "9")           # True! text comparison, not numeric
print(sorted(["10", "9", "100"]), sorted(["10", "9", "100"], key=int))
```

## 1. Changing case
`upper`, `lower`, `casefold`, `capitalize`, `title`, `swapcase`

```python
s = "hello WORLD, it's python"
print(s.upper())        # HELLO WORLD, IT'S PYTHON
print(s.lower())        # hello world, it's python
print(s.capitalize())   # Hello world, it's python - first letter only
print(s.title())        # Hello World, It'S Python - note the 'S bug
print(s.swapcase())     # HELLO world, IT'S PYTHON
```

```python
# casefold() is a stronger lower(), for comparing text in any language
print("Straße".lower(), "Straße".casefold())          # straße strasse
print("STRASSE".casefold() == "Straße".casefold())    # True

# real world: case-insensitive comparison
print("Yes".lower() in ["y", "yes"])
```

## 2. Removing whitespace and characters
`strip`, `lstrip`, `rstrip`, `removeprefix`, `removesuffix`

```python
raw = "   hello world  \n"
print(repr(raw.strip()))     # both ends
print(repr(raw.lstrip()))    # left only
print(repr(raw.rstrip()))    # right only

# with an argument: removes ANY of those characters from the ends
print("xxhixyx".strip("xy"))      # hi
print("$$$19.99".lstrip("$"))     # 19.99
```

```python
# strip() removes a SET of characters, which surprises people:
print("report.txt".rstrip(".txt"))       # 'repor' - the final t was removed too!

# removeprefix / removesuffix (Python 3.9+) remove an exact string, once
print("report.txt".removesuffix(".txt")) # report
print("Mr. Smith".removeprefix("Mr. "))  # Smith
print("Smith".removeprefix("Mr. "))      # Smith - unchanged if not there
```

## 3. Finding and counting
`find`, `rfind`, `index`, `rindex`, `count`, `startswith`, `endswith`

```python
s = "banana"
print(s.find("an"))      # 1  - first position
print(s.rfind("an"))     # 3  - last position
print(s.find("x"))       # -1 - not found
print(s.find("a", 2))    # 3  - start searching at index 2
print(s.count("a"))      # 3
print(s.count("ana"))    # 1  - matches don't overlap

# index / rindex work the same, but raise ValueError instead of returning -1
print(s.index("n"), s.rindex("n"))
try:
    s.index("x")
except ValueError as e:
    print("ValueError:", e)

# just need yes/no? use `in`
print("nan" in s)
```

```python
filename = "photo.JPG"
print(filename.startswith("photo"))
print(filename.lower().endswith((".jpg", ".png")))   # a tuple checks several options
```

## 4. Replacing
`replace`, `expandtabs`, `translate`, `maketrans`

```python
s = "one fish two fish red fish"
print(s.replace("fish", "cat"))       # every match
print(s.replace("fish", "cat", 1))    # only the first
print(s.replace(" ", ""))             # remove all spaces

print("a\tb\tc".expandtabs(4))        # tabs -> spaces
```

```python
# translate + maketrans: swap or delete many single characters in one go
table = str.maketrans("aeiou", "12345")
print("education".translate(table))          # 2d5c1t34n

table = str.maketrans("", "", "!?.,")        # third argument: characters to delete
print("Hi! How are you?".translate(table))   # Hi How are you

table = str.maketrans({"&": "and", "@": "at"})   # dict form: char -> any string
print("tea & cake @ 5".translate(table))
```

## 5. Splitting and joining
`split`, `rsplit`, `splitlines`, `partition`, `rpartition`, `join`

```python
print("a b   c".split())           # ['a', 'b', 'c'] - any whitespace, ignores extras
print("a,b,,c".split(","))         # ['a', 'b', '', 'c'] - keeps empty pieces
print("a,b,c,d".split(",", 2))     # ['a', 'b', 'c,d']   - at most 2 splits
print("a,b,c,d".rsplit(",", 1))    # ['a,b,c', 'd']      - split from the right
print("line1\nline2\r\nline3".splitlines())
```

```python
# partition: split into exactly 3 parts at the first match
print("key=value=x".partition("="))    # ('key', '=', 'value=x')
print("key=value=x".rpartition("="))   # ('key=value', '=', 'x')
print("novalue".partition("="))        # ('novalue', '', '') - never fails

key, _, value = "host=localhost".partition("=")
print(key, value)
```

```python
# join: the separator calls it, the list goes inside
words = ["Python", "is", "fun"]
print(" ".join(words))
print("-".join(words))
print("".join(["a", "b", "c"]))
print(", ".join(str(n) for n in [1, 2, 3]))   # items must be strings

try:
    ", ".join([1, 2, 3])
except TypeError as e:
    print("TypeError:", e)
```

## 6. Padding and aligning
`center`, `ljust`, `rjust`, `zfill`

```python
print("[" + "hi".center(10) + "]")
print("[" + "hi".center(10, "*") + "]")
print("[" + "hi".ljust(10, ".") + "]")
print("[" + "hi".rjust(10) + "]")
print("42".zfill(5), "-42".zfill(5))   # 00042 -0042 - keeps the sign in front
```

```python
# real world: a simple table
rows = [("Apple", 1.5), ("Banana", 0.25), ("Cherry", 12)]
for name, price in rows:
    print(name.ljust(10, ".") + f"${price:.2f}".rjust(8))
```

## 7. Checking what a string contains (`is...`)
All return `True` or `False`. An empty string returns `False` for all of them except `isascii` and `isprintable`.

`isalpha`, `isdigit`, `isdecimal`, `isnumeric`, `isalnum`, `isspace`, `islower`, `isupper`, `istitle`, `isidentifier`, `isprintable`, `isascii`

```python
tests = ["abc", "ABC", "Abc Def", "123", "abc123", "   ", "", "²", "½", "-5", "3.5", "my_var", "café", "a\tb"]
methods = ["isalpha", "isdigit", "isdecimal", "isnumeric", "isalnum", "isspace",
           "islower", "isupper", "istitle", "isidentifier", "isprintable", "isascii"]

print(f"{'':<10}" + "".join(m[2:6].ljust(6) for m in methods))
for t in tests:
    marks = "".join(("✓" if getattr(t, m)() else "·").ljust(6) for m in methods)
    print(f"{t!r:<10}{marks}")
```

```python
# isdecimal vs isdigit vs isnumeric: from strict to loose
for t in ["5", "²", "½"]:
    print(t, t.isdecimal(), t.isdigit(), t.isnumeric())

# use isdecimal() to check input before int() - but it rejects "-5" and "3.5"
for t in ["42", "-5", "3.5"]:
    print(t, t.isdecimal())
```

```python
# real world: safely convert user input to a number
def to_number(text):
    try:
        return float(text)
    except ValueError:
        return None

for t in ["42", "-5", "3.5", "abc"]:
    print(repr(t), "->", to_number(t))
```

## 8. Formatting
`format`, `format_map`

```python
print("{} is {} years old".format("John", 25))
print("{1} before {0}".format("A", "B"))                 # by position
print("{name} lives in {city}".format(name="Amy", city="Austin"))
print("{:>8.2f}|".format(3.14159))

person = {"name": "Raj", "city": "Dallas"}
print("{name} lives in {city}".format_map(person))        # straight from a dict
print("{name} lives in {city}".format(**person))          # same, using ** unpacking

# f-strings are usually easier today
name, age = "John", 25
print(f"{name} is {age} years old")
```

## 9. Encoding
`encode` turns text (`str`) into raw bytes (`bytes`). `bytes.decode()` goes back.

```python
text = "café €"
data = text.encode("utf-8")
print(data, len(text), len(data))       # 6 characters, 9 bytes
print(data.decode("utf-8"))

print("café".encode("ascii", errors="replace"))   # b'caf?' - can't store é in ASCII
```

### Clean up messy user input

### Make a URL slug from a title

```python
def slugify(title):
    keep = "".join(c if c.isalnum() else " " for c in title.lower())
    return "-".join(keep.split())

print(slugify("Hello, World! Learning Python in 2026"))
```

### Palindrome check

```python
def is_palindrome(text):
    letters = "".join(c for c in text.casefold() if c.isalnum())
    return letters == letters[::-1]

for t in ["racecar", "A man, a plan, a canal: Panama", "hello"]:
    print(f"{t!r}: {is_palindrome(t)}")
```

### Password strength checker

```python
def check_password(pw):
    problems = []
    if len(pw) < 8:
        problems.append("at least 8 characters")
    if not any(c.isupper() for c in pw):
        problems.append("an uppercase letter")
    if not any(c.islower() for c in pw):
        problems.append("a lowercase letter")
    if not any(c.isdigit() for c in pw):
        problems.append("a digit")
    if all(c.isalnum() for c in pw):
        problems.append("a symbol")
    return "strong" if not problems else "needs " + ", ".join(problems)

for pw in ["abc", "password123", "Passw0rd!"]:
    print(f"{pw:<12} {check_password(pw)}")
```

### Word frequency counter

```python
text = """The quick brown fox. The lazy dog! The fox, again?"""
table = str.maketrans("", "", ".,!?")
words = text.lower().translate(table).split()

counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1

for word, n in sorted(counts.items(), key=lambda kv: kv[1], reverse=True)[:3]:
    print(f"{word:<6} {n}")
```

### Mask sensitive data

```python
def mask_card(number):
    digits = number.replace(" ", "").replace("-", "")
    return digits[-4:].rjust(len(digits), "*")

def mask_email(email):
    user, _, domain = email.partition("@")
    return user[0] + "*" * (len(user) - 1) + "@" + domain

print(mask_card("4111 1111 1111 1234"))
print(mask_email("keerthi@example.com"))
```

### Parse a log line

```python
line = "2026-10-02 14:31:07 ERROR [payment] Card declined: insufficient funds"

date, time, level, rest = line.split(" ", 3)
module, _, message = rest.partition("] ")
module = module.removeprefix("[")

print(f"date:    {date}")
print(f"time:    {time}")
print(f"level:   {level}")
print(f"module:  {module}")
print(f"message: {message}")
```

### Parse a CSV line into a record

```python
header = "name,age,city"
row = "John Smith, 25 ,  Dallas"

record = dict(zip(header.split(","), (field.strip() for field in row.split(","))))
record["age"] = int(record["age"])
print(record)
```
