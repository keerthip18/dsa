# Built-in functions for numbers

Built-in functions are always available, with no `import` needed. `dir(__builtins__)` lists them all.

The ones for numbers fall into five groups:

| Group | Functions |
|---|---|
| Converting types | `int()`, `float()`, `complex()`, `bool()` |
| Math | `abs()`, `round()`, `pow()`, `divmod()` |
| Many numbers at once | `sum()`, `min()`, `max()`, `range()` |
| Other number bases | `bin()`, `oct()`, `hex()`, `int(text, base)` |
| Characters ↔ numbers | `ord()`, `chr()` |

Two helpers that work on any value come in handy too: `type()` and `isinstance()`, plus `format()` for display.

## 1. Converting types

### `int()`

```python
print(int(3.99))       # 3   - cuts off the decimals, does NOT round
print(int(-3.99))      # -3  - cuts toward zero
print(int("42"))       # 42  - from a string
print(int("  42\n"))   # 42  - surrounding spaces are ignored
print(int("1_000"))    # 1000 - underscores allowed, like in code
print(int(True))       # 1
print(int())           # 0
```

```python
# strings must look like whole numbers
for text in ["3.5", "abc", ""]:
    try:
        int(text)
    except ValueError as e:
        print("ValueError:", e)

# to turn "3.5" into an int, go through float first
print(int(float("3.5")))   # 3
```

### `float()`

```python
print(float(7))         # 7.0
print(float("3.14"))    # 3.14
print(float("1e3"))     # 1000.0 - scientific notation
print(float("inf"), float("-inf"))   # infinity
print(float("nan"))     # "not a number"

# nan is never equal to anything, not even itself
nan = float("nan")
print(nan == nan)       # False
```

```python
# floats are stored in binary, so some decimals can't be exact
print(0.1 + 0.2)               # 0.30000000000000004
print(0.1 + 0.2 == 0.3)        # False!
print(round(0.1 + 0.2, 2) == 0.3)   # True - round before comparing

import math
print(math.isclose(0.1 + 0.2, 0.3))  # True - the proper way
```

### `complex()`

```python
z = complex(2, 3)       # 2 + 3j
print(z, z.real, z.imag)
print(complex("1+2j"))  # from a string (no spaces allowed)
print(abs(complex(3, 4)))   # 5.0 - distance from 0
```

### `bool()`

```python
# for numbers: 0 is False, everything else is True
for n in [0, 0.0, 1, -5, 0.001, float("nan")]:
    print(f"{n!r:<6} -> {bool(n)}")

# bool is a subclass of int, so True and False act like 1 and 0
print(True + True)          # 2
print(sum([True, False, True, True]))   # 3 - handy for counting
```

### Checking types: `type()` and `isinstance()`

```python
print(type(5), type(5.0), type(5j), type(True))

# isinstance is usually better: it accepts a tuple of types
for value in [5, 5.0, "5", True]:
    print(f"{value!r:<5} is a number? {isinstance(value, (int, float))}")

# careful: True counts as an int, because bool is a subclass of int
print(isinstance(True, int))   # True
```

## 2. Math

### `abs()`: distance from zero

```python
print(abs(-7), abs(7), abs(-2.5), abs(0))

# real world: how far apart are two values, regardless of order?
target, guess = 50, 38
print(f"off by {abs(target - guess)}")
```

### `round()`

```python
print(round(3.14159, 2))   # 3.14
print(round(3.7))          # 4    - no digits -> returns an int
print(round(3.7, 0))       # 4.0  - with digits -> returns a float
print(round(1234, -2))     # 1200 - negative digits round to tens, hundreds...
print(round(1250, -2))     # 1200 - see the next cell for why not 1300
```

```python
# halves round to the nearest EVEN number ("banker's rounding")
print(round(0.5), round(1.5), round(2.5), round(3.5))   # 0 2 2 4

# and some halves aren't really halves, because of binary storage
print(round(2.675, 2))     # 2.67, not 2.68

# for money, use the decimal module to round the way people expect
from decimal import Decimal, ROUND_HALF_UP
print(Decimal("2.675").quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))   # 2.68
```

```python
# rounding for display vs. rounding the value
price = 19.98765
print(round(price, 2))     # changes the number
print(f"{price:.2f}")      # only changes how it's shown
```

### `pow()`

```python
print(pow(2, 10))       # 1024 - same as 2 ** 10
print(pow(2, -1))       # 0.5
print(pow(9, 0.5))      # 3.0 - square root

# the 3-argument form: (base ** exp) % mod, but much faster for big numbers
print(pow(3, 200, 7))
print(3 ** 200 % 7)     # same answer, but builds a huge number first
```

### `divmod()`: quotient and remainder together

```python
q, r = divmod(17, 5)
print(q, r)                         # 3 2  - same as (17 // 5, 17 % 5)

# real world: seconds -> hours, minutes, seconds
total = 7384
minutes, seconds = divmod(total, 60)
hours, minutes = divmod(minutes, 60)
print(f"{hours}h {minutes}m {seconds}s")   # 2h 3m 4s
```

```python
# real world: how many boxes, and how many items left over?
items, per_box = 50, 12
boxes, left = divmod(items, per_box)
print(f"{boxes} full boxes, {left} items left over")
```

## 3. Working with many numbers

### `sum()`

```python
nums = [4, 8, 15, 16, 23, 42]
print(sum(nums))                 # 108
print(sum(nums, 100))            # 208 - start value
print(sum(n * n for n in nums))  # sum of squares, using a generator
print(sum([]))                   # 0 - empty is fine

# floats: since Python 3.12, sum() corrects rounding errors as it adds
total = 0
for x in [0.1] * 10:
    total += x
print(total)                 # 0.9999999999999999 - plain loop
print(sum([0.1] * 10))       # 1.0
```

### `min()` and `max()`

```python
nums = [4, 8, 15, 16, 23, 42]
print(min(nums), max(nums))      # from an iterable
print(min(3, 1, 2), max(3, 1, 2))  # or from separate arguments

# empty input raises an error, unless you give a default
print(max([], default=0))
```

```python
# key= picks WHAT to compare, and still returns the original item
temps = [-8, 3, -15, 12]
print(max(temps, key=abs))       # -15 - furthest from zero

players = [("Amy", 820), ("John", 950), ("Raj", 710)]
print(max(players, key=lambda p: p[1]))   # ('John', 950)
```

```python
# real world: clamp a value into a range
def clamp(value, low, high):
    return max(low, min(value, high))

print(clamp(150, 0, 100))   # 100 - volume can't go above 100
print(clamp(-20, 0, 100))   # 0
print(clamp(42, 0, 100))    # 42
```

### `range()`

```python
print(list(range(5)))          # [0, 1, 2, 3, 4]       - stop
print(list(range(2, 7)))       # [2, 3, 4, 5, 6]       - start, stop
print(list(range(0, 20, 5)))   # [0, 5, 10, 15]        - start, stop, step
print(list(range(10, 0, -3)))  # [10, 7, 4, 1]         - counting down

# range is lazy and tiny in memory, however big it is
big = range(1_000_000_000)
print(len(big), big[-1], 999_999 in big)
```

## 4. Other number bases

```python
n = 42
print(bin(n))   # 0b101010  - base 2
print(oct(n))   # 0o52      - base 8
print(hex(n))   # 0x2a      - base 16

# without the prefix, use format()
print(format(n, "b"), format(n, "o"), format(n, "x"), format(n, "08b"))
```

```python
# int(text, base) goes the other way
print(int("101010", 2))   # 42
print(int("52", 8))       # 42
print(int("2a", 16))      # 42
print(int("0x2a", 0))     # 42 - base 0 reads the prefix to decide

# you can also write numbers in these bases directly in code
print(0b101010, 0o52, 0x2a)
```

```python
# real world: hex color code -> RGB values
color = "#1E90FF"
r = int(color[1:3], 16)
g = int(color[3:5], 16)
b = int(color[5:7], 16)
print(r, g, b)                                  # 30 144 255
print(f"#{r:02X}{g:02X}{b:02X}")                # back to hex
```

## 5. Characters ↔ numbers: `ord()` and `chr()`

```python
print(ord("A"), ord("a"), ord("0"))   # 65 97 48 - each character has a number
print(chr(65), chr(97), chr(8364))    # A a €

# all lowercase letters
print("".join(chr(c) for c in range(ord("a"), ord("z") + 1)))
```

```python
# real world: Caesar cipher - shift each letter 3 places
def caesar(text, shift):
    result = []
    for ch in text:
        if ch.isalpha() and ch.islower():
            result.append(chr((ord(ch) - ord("a") + shift) % 26 + ord("a")))
        else:
            result.append(ch)
    return "".join(result)

secret = caesar("hello world", 3)
print(secret)                # khoor zruog
print(caesar(secret, -3))    # hello world
```

## 6. Putting it together

```python
# a bill splitter
bill = 187.40
tip_percent = 18
people = 4

tip = round(bill * tip_percent / 100, 2)
total = bill + tip
each = round(total / people, 2)

print(f"bill:  ${bill:>8.2f}")
print(f"tip:   ${tip:>8.2f}  ({tip_percent}%)")
print(f"total: ${total:>8.2f}")
print(f"each:  ${each:>8.2f}  x {people}")
```

```python
# a stats summary from raw text input
raw = "72, 88, 95, 61, 79, 88, 100"
scores = [int(s) for s in raw.split(",")]

print(f"count:   {len(scores)}")
print(f"total:   {sum(scores)}")
print(f"average: {round(sum(scores) / len(scores), 1)}")
print(f"range:   {min(scores)} to {max(scores)} (spread {max(scores) - min(scores)})")
print(f"passed:  {sum(s >= 70 for s in scores)}")   # True counts as 1
```
