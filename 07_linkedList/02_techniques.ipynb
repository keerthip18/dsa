```python
class ListNode:

    def __init__(self, val, next_node = None):
        self.val = val
        self.next = next_node

    def __str__(self):
        string = ""
        curr = self
        while curr:
            string += str(curr.val)
            if(curr.next):
                string += " -> "
            curr = curr.next
        return string
```

```python
#build linked list
def build_list(list):
    head = None
    tail = None
    for num in list:
        node = ListNode(num)

        if head is None:
            head = node 

        else: 
            tail.next = node

        tail = node

    return head

list = [10, 20, 30]
head = build_list(list)
print(head)

#build cyclic linked list
def build_cyclic_list(list, cycle_at = 0):
    head = build_list(list)
    curr = head
    cycle_node = None
    count = 0

    while curr:
        tail = curr

        if count == cycle_at:
            cycle_node = curr

        count += 1
        curr = curr.next
        
    tail.next = cycle_node
    return head

list = [10, 20, 30]
head = build_cyclic_list(list, 2)
print(head)
```

```python
#reverse linked list
def reverse(head):
    curr = head 
    prev = None 

    while curr:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node

    return prev

prev = reverse(head)
print(prev)
```

```python
#middle of linked list
def middle(head):
    slow = head 
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next 
    return slow

list = [10, 20, 30, 40, 50, 60]
head = build_list(list)
mid_node = middle(head)
print(mid_node.val)
```

```python
def has_cycle(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            return True

    return False

list = [10, 20, 30, 40, 50, 60]
head = build_cyclic_list(list, 2)
cycle = has_cycle(head)
print(cycle)
```
