```python
class ListNode:

    def __init__(self, val, next_node = None):
        self.val = val
        self.next = next_node

    def __str__(self):
        string = ""
        curr = self
        while curr:
            string += str(curr.val)
            if(curr.next):
                string += " -> "
            curr = curr.next
        return string
```

```python
#build linked list
def build_list(list):
    head = None
    tail = None
    for num in list:
        node = ListNode(num)

        if head is None:
            head = node 

        else: 
            tail.next = node

        tail = node

    return head

list = [10, 20, 30]
head = build_list(list)
print(head)
```

```python
#identify the kth node from the end
def find_kth_node_end(head, k):
    #start fast k ahead and then move each by one
    slow = head 
    fast = head
    if k > 1:
        for _ in range(k):
            if fast is None:
                return None  #k is larger than the list length
            fast = fast.next
        while fast:
            slow = slow.next
            fast = fast.next
        return slow
    return None
list = [10, 20, 30, 40, 50]
head = build_list(list)
kth_node = find_kth_node_end(head, 2)
if kth_node:
    print(kth_node.val)
```

```python
#insert a new node at the kth position from end 
def insert_at_kth_node_end(head, node, k):
    if k > 1:
        kth_node = find_kth_node_end(head, k)
        temp = kth_node.next 
        kth_node.next = node 
        node.next = temp
        print(head)

node = ListNode(30)
insert_at_kth_node_end(head, node, 2)
```

```python
#delete a node at the kth position from end
def delete_at_kth_node_end(head, k):
    if k > 1:
        kth_node = find_kth_node_end(head, k)
        prev = head
        while prev:
            prev = prev.next
            if prev.next is kth_node:
                break
        prev.next = kth_node.next
        print(head)

delete_at_kth_node_end(head, 2)   
```

```python
#update a node at the kth position from end
def update_at_kth_node_end(head, new_val, k):
    if k > 1:
        kth_node = find_kth_node_end(head, k)
        kth_node.val = new_val 
        print(head)

update_at_kth_node_end(head, 50, 2)
```
