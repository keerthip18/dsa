```python
class ListNode:

    def __init__(self, val, next_node = None):
        self.val = val
        self.next = next_node

    def __str__(self):
        string = ""
        curr = self
        while curr:
            string += str(curr.val)
            if(curr.next):
                string += " -> "
            curr = curr.next
        return string
```

```python
#build linked list
def build_list(list):
    head = None
    tail = None
    for num in list:
        node = ListNode(num)

        if head is None:
            head = node 

        else: 
            tail.next = node

        tail = node

    return head

list = [10, 20, 30]
head = build_list(list)
print(head)

#build cyclic linked list
def build_cyclic_list(list, cycle_at = 0):
    head = build_list(list)
    curr = head
    cycle_node = None
    count = 0

    while curr:
        tail = curr

        if count == cycle_at:
            cycle_node = curr

        count += 1
        curr = curr.next
        
    tail.next = cycle_node
    return head

list = [10, 20, 30]
head = build_cyclic_list(list, 2)
#print(head)
```

```python
def has_cycle(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            return True

    return False

list = [10, 20, 30, 40, 50, 60]
head = build_cyclic_list(list, 2)
cycle = has_cycle(head)
print(cycle)
```

```python
#find start of cycle
def cycle_start(head):
    #after determining cycle, reset slow to head and move both pointers by one and will meet at start of cycle
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            break
    else:
        return -1 #no cycle

    slow = head
    index = 0
    while slow is not fast:
        slow = slow.next
        fast = fast.next
        index += 1

    return index

list = [10, 20, 30, 40, 50, 60]
head = build_cyclic_list(list, 50)
cycle_at = cycle_start(head)
print(cycle_at)
```
