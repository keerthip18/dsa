```python
class ListNode:
    def __init__(self, val, next_node = None):
        self.value = val
        self.next = next_node

    def __str__(self):
        string = ""
        current = self
        while current:
            string += str(current.value)
            if(current.next):
                string += " -> "
            current = current.next
        return string

#build linked list
def build_list(list):
    head = None
    tail = None
    for num in list:
        node = ListNode(num)

        if head is None:
            head = node 

        else: 
            tail.next = node

        tail = node

    return head

list = [10, 20, 30]
head = build_list(list)
print(head)
```

```python
def traverse(head):
    curr = head 
    count = 0
    while curr and curr.next:
        start = curr
        end = curr.next
        print(f"edge {count}: {start.value}, {end.value}")
        curr = curr.next
        count += 1

    print(f"end of the loop curr: {curr}")
    print(f"end of the loop count: {count}")


traverse(head)
```

```python
def adjacent_swap(head):
    dummy = ListNode(0, head)
    prev = dummy # node before the current pair
    while prev.next and prev.next.next:
        first = prev.next # 10
        second = first.next # 20
        third = second.next # 30
        second.next = first
        first.next = third
        prev.next = second # link previous part to the new pair head
        prev = first # first is now the tail of the swapped pair
    return dummy.next

list = [10, 20, 30, 40]
head = build_list(list)
swapped = adjacent_swap(head)
print(swapped)
```
