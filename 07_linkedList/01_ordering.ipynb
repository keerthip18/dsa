# Linked List — Ordering Techniques

| # | Technique | Pattern |
|---|-----------|---------|
| 1 | Traversal to print edges | Single pointer walk |
| 2 | Find the middle node | Two pointers (slow / fast) |
| 3 | Swap adjacent nodes | Pairwise pointer rewiring |
| 4 | Reverse a linked list | `prev` / `curr` / `next` |
| 5 | Merge two sorted lists | Dummy head + two pointers |
| 6 | Odd–even list | Split into two chains, then join |
| 7 | Reorder list: first, last, second, second-last, … | Find middle → reverse second half → merge alternately |

```python
class ListNode:
    def __init__(self, val, next_node = None):
        self.value = val
        self.next = next_node

    def __str__(self):
        string = ""
        current = self
        while current:
            string += str(current.value)
            if(current.next):
                string += " -> "
            current = current.next
        return string

#build linked list
def build_list(list):
    head = None
    tail = None
    for num in list:
        node = ListNode(num)

        if head is None:
            head = node 

        else: 
            tail.next = node

        tail = node

    return head

list = [10, 20, 30]
head = build_list(list)
print(head)
```

```python
def traverse(head):
    curr = head 
    count = 0
    while curr and curr.next:
        start = curr
        end = curr.next
        print(f"edge {count}: {start.value}, {end.value}")
        curr = curr.next
        count += 1

    print(f"end of the loop curr: {curr}")
    print(f"end of the loop count: {count}")


traverse(head)
```

```python
def adjacent_swap(head):
    dummy = ListNode(0, head)
    prev = dummy
    while prev.next and prev.next.next:
        first = prev.next #10
        second = first.next #20
        first.next = second.next
        second.next = first
        prev.next = second
        prev = first
    return dummy.next

list = [10, 20, 30, 40]
head = build_list(list)
swapped = adjacent_swap(head)
print(swapped)
```

```python
def odd_even_list(head):
    if head is None:
        return head
    
    odd_list = head
    even_list = head.next
    even_head = even_list
   
    while even_list and even_list.next: #dont need to identify even or odd node using division/modulus
        odd_list.next = even_list.next
        odd_list = odd_list.next #move odd pointer first or else infinite loop
        even_list.next = odd_list.next
        even_list = even_list.next
    
    odd_list.next = even_head
    return head


list = [10, 20, 30, 40, 50, 60]
head = build_list(list)
result = odd_even_list(head)
print(result)
```

```python
#reverse linked list
def reverse(head):
    curr = head 
    prev = None 

    while curr:
        temp = curr.next
        curr.next = prev
        prev = curr
        curr = temp

    return prev

prev = reverse(head)
print(prev)
```

```python
#middle of linked list
def middle(head):
    slow = head 
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next 
    return slow

list = [10, 20, 30, 40, 50, 60]
head = build_list(list)
mid_node = middle(head)
print(mid_node.value)
```

```python
#reorder linked list: first, last, second, second last ...
def reorder(head):
    if head is None or head.next is None:
        return head

    #split at middle
    mid = middle(head)
    second = mid.next
    mid.next = None

    #reverse second half
    second = reverse(second)

    #merge alternately
    first = head
    while second:
        first_next = first.next
        second_next = second.next

        first.next = second
        second.next = first_next

        first = first_next
        second = second_next

    return head

list = [10, 20, 30, 40, 50]
head = build_list(list)
head = reorder(head)
print(head)
```

```python
#merge two sorted linked lists 
def merge(list1, list2):
    result_list = ListNode(0)
    tail = result_list
    while list1 and list2:
        if list1.value < list2.value:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next
        tail = tail.next
    tail.next = list1 or list2
    return result_list.next

list1 = [1, 3, 5, 7]
linked_list1 = build_list(list1)
list2 = [2, 4, 6, 10]
linked_list2 = build_list(list2)
result = merge(linked_list1, linked_list2)
print(result)
```
