```python
class ListNode:
    def __init__(self, val, next = None):
        self.value = val
        self.nextNode = next

    def __str__(self):
        string = ""
        current = self
        while current:
            string += str(current.value)
            if(current.nextNode):
                string += " -> "
            current = current.nextNode
        return string

    def append(self, val):
        node = ListNode(val)
        current = self
        while current.nextNode:
            current = current.nextNode
        current.nextNode = node


root = ListNode(10)
print(root)
root.append(20)
root.append(30)
root.append(40)
print(root)
    
```

```python
def linkedList(list):
    root = tail = None
    for num in list:
        node = ListNode(num)
        if root is None:
            root = node
        else:
            tail.nextNode = node
        tail = node
    return root

root = linkedList([10, 20, 30, 40])
print(root)
```

```python
def cyclicList(list):
    root = tail = None
    for num in list:
        node = ListNode(num)
        if root is None:
            root = node
        else:
            tail.nextNode = node
        tail = node
    tail.nextNode = root
    return root

root = cyclicList([10, 20, 50, 60])

def hasCycle(root):
    slow = root
    fast = root 
    while fast:
        if fast.nextNode == slow:
            return True
        else:
            fast = fast.nextNode
    return False

cycle1 = hasCycle(root)
root = linkedList([10, 20, 50, 60])
cycle2 = hasCycle(root)
print(cycle2)
```
