```python
class Student:
    def __init__(self, num, name, age):
        self.number = num
        self.name = name
        self.age = age
        self.grade = 0

    def __str__(self):
        # Friendly text for end users: used by print() and str()
        return f"Student {self.name} with number {self.number} is {self.age} years old and grade is {self.grade}"
    
    def __repr__(obj):
        # Unambiguous text for developers: used in the REPL, debugger, and inside lists
        return f"Student(num={obj.number!r}, name={obj.name!r}, age={obj.age!r})"

    def set_grade(self, num):
        self.grade = num

    def get_grade(self):
        return self.grade

    def curve_grade(self):
        self.grade += 10

student1 = Student(1, "Keerthi", 21)
student2 = Student(2, "Doomp", 15)
student1.name = "Keerthi Padavala"
student1.set_grade(90)
student1.curve_grade()

print(student1)          # calls __str__
```

```python
class SpecialStudent(Student):
    def __str__(self):
            # Friendly text for end users: used by print() and str()
            str = super().__str__()
            return f"{str} with honors subject {self.honorsSubject}"
    
    def honors(self, subject):
        self.honorsSubject = subject

student3 = SpecialStudent(3, "milo", 2)
student3.honorsSubject = "dog training"
print(student3)
```
