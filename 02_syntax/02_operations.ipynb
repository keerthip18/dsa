```python
# int operations
a = 10
b = 3
print(a + b)   # addition
print(a - b)   # subtraction
print(a * b)   # multiplication
print(a / b)   # true division -> float
print(a // b)  # floor division
print(a % b)   # modulus
print(a ** b)  # exponent
```

```python
# float operations
x = 5.5
y = 2.0
print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x // y)
print(x % y)
print(round(x, 0))
```

```python
# complex operations
c1 = 2 + 3j
c2 = 1 - 1j
print(c1 + c2)
print(c1 - c2)
print(c1 * c2)
print(c1 / c2)
print(c1.real, c1.imag)
print(abs(c1))
```

```python
# bool operations
p = True
q = False
print(p and q)
print(p or q)
print(not p)
print(p == q)
print(p + p)   # bool behaves like int (True == 1)
```

```python
# str operations
s1 = "Hello"
s2 = "World"
print(s1 + " " + s2)   # concatenation
print(s1 * 3)          # repetition
print(s1[1:4])         # slicing
print(s1.upper())
print(s1.lower())
print(len(s1))
print(s1 in "Hello there")
```

```python
# bytes operations
b1 = b"hello"
b2 = b"world"
print(b1 + b2)          # concatenation
print(b1 * 2)           # repetition
print(b1[1:4])          # slicing
print(b1.decode())      # bytes -> str
print(len(b1))
```

```python
# NoneType operations
n = None
print(n is None)
print(n is not None)
print(n == None)
```

```python
# ternary operator: value_if_true if condition else value_if_false
age = 20
status = "adult" if age >= 18 else "minor"
print(status)

a, b = 10, 3
print(a if a > b else b)   # max of two numbers

n = 7
print("even" if n % 2 == 0 else "odd")

# chained ternary (use sparingly - hard to read)
score = 75
grade = "A" if score >= 90 else "B" if score >= 70 else "C"
print(grade)
```
