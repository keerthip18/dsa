```python
list = [1, 2, 3, 4]

def minimum(list):
    minVal = list[0]
    for num in list:
        if num < minVal:
            minVal = num
    return minVal

print(f"min: {minimum(list)}")

def maximum(list):
    maxVal = list[0]
    for num in list:
        if num > maxVal:
            maxVal = num 
    return maxVal

print(f"max: {maximum(list)}")

def summation(list):
    sum = 0
    for num in list:
        sum = sum + num
    return sum

print(f"sum: {summation(list)}")
 

def length(list):
    count = 0
    for _ in list:
        count = count + 1
    return count

print(f"length : {length(list)}") 
```

```python
import heapq

nums = [7, 2, 9, 4, 1, 8]

# min heap: heapq keeps the smallest item at index 0
minHeap = nums[:]         
heapq.heapify(minHeap)     

print(f"min: {minHeap[0]}")  

# max heap: heapq only does min heaps, so store negatives
maxHeap = [-num for num in nums]
heapq.heapify(maxHeap)     

print(f"max: {-maxHeap[0]}") 

# the real win: min/max stay O(1) as values change, with O(log n) updates
heapq.heappush(minHeap, 0)       
heapq.heappush(maxHeap, -20)    
print(f"after push -> min: {minHeap[0]}, max: {-maxHeap[0]}")

heapq.heappop(minHeap)           
heapq.heappop(maxHeap)          
print(f"after pop  -> min: {minHeap[0]}, max: {-maxHeap[0]}")
```
