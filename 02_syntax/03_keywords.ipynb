```python
# 1. Boolean and None literals: True, False, None
is_ready = True
is_done = False
result = None
print(is_ready, is_done, result)
```

```python
# 2. Logical operators: and, or, not
a, b = True, False
print(a and b)
print(a or b)
print(not a)
```

```python
# 3. Identity and membership: is, in, not
x = None
nums = [1, 2, 3]
print(x is None)
print(x is not None)
print(2 in nums)
print(5 not in nums)
```

```python
# 4. Conditionals: if, elif, else
score = 75
if score >= 90:
    print("A")
elif score >= 60:
    print("B")
else:
    print("C")
```

```python
# 5. Loops: for, while, break, continue, pass
for n in range(5):
    if n == 2:
        continue   # skip this iteration
    if n == 4:
        break      # stop the loop
    print("for:", n)

count = 0
while count < 3:
    if count == 1:
        pass       # placeholder, does nothing
    print("while:", count)
    count += 1
```

```python
# 6. Functions: def, return, lambda
def square(x):            # 'def' defines a function
    return x * x          # 'return' sends back a value

double = lambda x: x * 2  # 'lambda' makes a small anonymous function
print(square(4))
print(double(4))
```

```python
# 7. Scope: global, nonlocal
counter = 0

def increment_global():
    global counter    # modify the outer/global variable
    counter += 1

def outer():
    total = 0
    def inner():
        nonlocal total   # modify the enclosing function's variable
        total += 10
    inner()
    return total

increment_global()
increment_global()
print("counter:", counter)
print("outer total:", outer())
```

```python
# 8. Exceptions: try, except, finally, raise, assert
def divide(a, b):
    assert b != 0, "b must not be zero"  # 'assert' checks a condition
    return a / b

try:
    print(divide(10, 2))
    print(divide(10, 0))
except ZeroDivisionError:
    print("caught divide by zero")
except AssertionError as e:
    print("assertion failed:", e)
finally:
    print("cleanup done")

try:
    raise ValueError("custom error")   # 'raise' triggers an exception
except ValueError as e:
    print("caught:", e)
```

```python
# 9. Classes and objects: class, del
class Dog:                 # 'class' defines a new type
    def __init__(self, name):
        self.name = name

    def bark(self):
        return f"{self.name} says woof!"

d = Dog("Rex")
print(d.bark())

del d.name                 # 'del' removes an attribute/reference
print(hasattr(d, "name"))
```

```python
# 10. Modules: import, from, as
import math                            # 'import' brings in a module
from math import sqrt as square_root   # 'from', 'as'

print(math.pi)
print(square_root(16))
```

```python
# 11. Context managers: with
with open("/tmp/kw_with_demo.txt", "w") as f:   # 'with' + 'as'
    f.write("hello")

with open("/tmp/kw_with_demo.txt") as f:
    print(f.read())
```

```python
# 12. Generators: yield
def countdown(n):
    while n > 0:
        yield n       # 'yield' pauses and produces a value
        n -= 1

for value in countdown(3):
    print(value)
```

```python
# 13. Async: async, await
import asyncio

async def fetch_data():      # 'async' defines a coroutine
    await asyncio.sleep(0)   # 'await' pauses until the task finishes
    return "data ready"

result = asyncio.run(fetch_data())
print(result)
```

```python
# 14. Soft keywords: match, case, _
def describe(value):
    match value:               # 'match' starts pattern matching
        case 0:
            return "zero"
        case int():
            return "an int"
        case _:                 # '_' is the wildcard/default case
            return "something else"

print(describe(0))
print(describe(5))
print(describe("hi"))
```

```python
# 15. Soft keyword: type (type alias, Python 3.12+)
type Vector = list[float]   # 'type' creates a type alias

v: Vector = [1.0, 2.0, 3.0]
print(v)
```
