```python
# 1. Basic for loop over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# 2. for with range()
for i in range(5):
    print(i)
```

```python
# 3. for with range(start, stop, step)
for i in range(10, 0, -2):
    print(i)
```

```python
# 4. for over a string (iterates characters)
for ch in "Py":
    print(ch)
```

```python
# 5. for over a dict: keys, values, items
student = {"name": "Alice", "age": 20}
for key in student:
    print("key:", key)
for value in student.values():
    print("value:", value)
for key, value in student.items():
    print("item:", key, value)
```

```python
# 6. for with enumerate() -> index + value
colors = ["red", "green", "blue"]
for index, color in enumerate(colors):
    print(index, color)

for index, color in enumerate(colors, start=1):  # custom start
    print(index, color)
```

```python
# 7. for with zip() -> loop over multiple sequences together
names = ["Alice", "Bob"]
ages = [30, 25]
for name, age in zip(names, ages):
    print(name, age)
```

```python
# 8. for...else -> else runs only if loop completes without 'break'
for n in range(3):
    print("checking", n)
else:
    print("loop finished normally")

for n in range(3):
    if n == 1:
        break
    print("checking", n)
else:
    print("this will NOT print")
```

```python
# 9. Nested for loops
for i in range(2):
    for j in range(2):
        print(f"i={i}, j={j}")
```

```python
# 10. for with break and continue
for n in range(6):
    if n == 4:
        break        # stop the loop entirely
    if n % 2 == 0:
        continue     # skip even numbers
    print(n)
```

```python
# 11. List comprehension (for inside [])
squares = [x * x for x in range(5)]
print(squares)

evens = [x for x in range(10) if x % 2 == 0]   # with condition
print(evens)
```

```python
# 12. Dict comprehension (for inside {})
squares_map = {x: x * x for x in range(5)}
print(squares_map)
```

```python
# 13. Set comprehension (for inside {})
unique_lengths = {len(word) for word in ["hi", "bye", "ok", "yo"]}
print(unique_lengths)
```

```python
# 14. Generator expression (for inside ())
gen = (x * 2 for x in range(5))
print(type(gen))      # generator object, values computed lazily
print(list(gen))
```

```python
# 15. for over reversed() and sorted()
nums = [3, 1, 2]
for n in reversed(nums):
    print("reversed:", n)
for n in sorted(nums):
    print("sorted:", n)
```

```python
# 16. Unpacking tuples directly in a for loop
points = [(0, 0), (1, 2), (3, 4)]
for x, y in points:
    print(f"x={x}, y={y}")
```

```python
# 17. Nested comprehension (flatten a 2D list)
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [value for row in matrix for value in row]
print(flat)
```
