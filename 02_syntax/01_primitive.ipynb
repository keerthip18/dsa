```python
age = 25
print(age, type(age))  # 25 <class 'int'>
```

```python
height = 5.9
print(height, type(height))  # 5.9 <class 'float'>
```

```python
impedance = 3 + 4j
print(impedance, type(impedance))  # (3+4j) <class 'complex'>
```

```python
is_active = True
print(is_active, type(is_active))  # True <class 'bool'>
```

```python
name = "Keerthi"
print(name, type(name))  # Keerthi <class 'str'>
```

```python
data = b"hello"
print(data, type(data))  # b'hello' <class 'bytes'>
```

```python
result = None
print(result, type(result))  # None <class 'NoneType'>
```
