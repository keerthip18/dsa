```python
# 1. Basic f-string interpolation
name = "Alice"
age = 30
print(f"Name: {name}")
print(f"{name} is {age} years old")
```

```python
# 2. Expressions, function calls, and method calls inside f-strings
x, y = 4, 5
name = "bob"
print(f"Sum: {x + y}")
print(f"Upper: {name.upper()}")
print(f"Max: {max(3, 7, 2)}")
```

```python
# 3. Width, alignment, and fill character
word = "hi"
print(f"[{word:10}]")     # default left-align for strings, width 10
print(f"[{word:>10}]")    # right-align
print(f"[{word:^10}]")    # center-align
print(f"[{word:*^10}]")   # center-align with '*' fill
```

```python
# 4. Precision for floats
pi = 3.14159265
print(f"{pi:.2f}")     # 2 decimal places
print(f"{pi:.4f}")     # 4 decimal places
print(f"{pi:10.2f}")   # width 10, 2 decimals
```

```python
# 5. Zero-padding and sign
num = 42
print(f"{num:05d}")    # zero-padded to width 5
print(f"{num:+d}")     # force a sign
print(f"{-num:+d}")
```

```python
# 6. Thousands separator and percentage
big = 1234567
ratio = 0.4567
print(f"{big:,}")        # comma separator
print(f"{ratio:.2%}")    # percentage with 2 decimals
```

```python
# 7. Hex, octal, and binary
n = 255
print(f"{n:x}")   # hex
print(f"{n:o}")   # octal
print(f"{n:b}")   # binary
print(f"{n:#x}")  # hex with 0x prefix
```

```python
# 8. Conversion flags: !r, !s, !a
text = "hi\tthere"
print(f"{text!r}")   # repr() - shows quotes/escapes
print(f"{text!s}")   # str() - plain string
print(f"{text!a}")   # ascii() - escapes non-ascii
```

```python
# 9. Debug specifier: f"{expr=}"
value = 42
print(f"{value=}")
print(f"{value * 2=}")
```

```python
# 10. Nested fields: dynamic width/precision from variables
width = 10
precision = 3
pi = 3.14159265
print(f"{pi:{width}.{precision}f}")
```

```python
# 11. Multi-line f-string
name = "Alice"
age = 30
print(f"""
Name: {name}
Age: {age}
""")
```
