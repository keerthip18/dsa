```python
from collections import deque
q = deque()
q.append(10)
q.append(20)
q.append(30)
print(q)

while q:
    e = q.popleft()
    print(e)

print(q)
```

```python
def is_palindrome(word):
    l_word = word.lower()
    q = deque(l_word)

    while q and len(q) >= 2:
        if q.pop() != q.popleft():
            return False
    
    return True

bool = is_palindrome("tt")
print(bool)
```
