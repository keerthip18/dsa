```python
#list traversal from the beginning
list = [10, 20, 30]
for i in range(len(list)):
    print(list[i])
```

```python
#list traversal from the end
list = [10, 20, 30]
for i in range(len(list) - 1, -1, -1):
    print(list[i])
```

```python
stack = []
stack.append(10) 
stack.append(20)
stack.append("hello")
print(stack)

while stack:
    e = stack.pop()
    print(e)

print(stack)
```
