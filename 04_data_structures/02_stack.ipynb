```python
def next_greater(nums):
    result = [-1] * len(nums)
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):   
            if nums[j] > nums[i]:
                result[i] = nums[j]      
                break                      
    return result

nums = [70, 65, 68, 72, 66]
result = next_greater(nums)
print(result)
```

```python
def next_greater_days(nums):
    result = [-1] * len(nums)
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):   
            if nums[j] > nums[i]:
                result[i] = j - i     
                break                      
    return result

nums = [70, 65, 68, 72, 66]
result = next_greater_days(nums)
print(result)
```

```python
def prev_greater(nums):
    result = [-1] * len(nums)
    for i in range(len(nums)):
        for j in range(i - 1, -1, -1):   
            if nums[j] > nums[i]:
                result[i] = nums[j]      
                break                      
    return result

nums = [1, 4, 3]
result = prev_greater(nums)
print(result)
```

```python
def stack_traverse(nums):
    stack = []
    for num in nums:
        stack.append(num)
    while stack:
        e = stack.pop()
        print(e)

stack_traverse(nums)
```

```python
def next_greater_stack(nums):
    result = [-1] * len(nums)
    stack = []
    stack.append((nums[0], 0))
    for index, num in enumerate(nums):
        val_stack, idx_stack = stack[-1]
        if num < val_stack:
            stack.append((num, index))
        else:
            while stack and stack[-1][0] < num:
                val2_stack, idx2_stack = stack.pop()
                if num > val2_stack:
                    result[idx2_stack] = num
            stack.append((num, index))
    return result

nums = [70, 65, 68, 72, 66]
print(next_greater_stack(nums))
```

```python
def next_greater_stack_short(nums):
    result = [-1] * len(nums)
    stack = []
    for index, num in enumerate(nums):
        while stack and stack[-1][0] < num:
            val_stack, idx_stack = stack.pop()
            result[idx_stack] = num
        stack.append((num, index))
    return result

nums = [70, 65, 68, 72, 66]
print(next_greater_stack_short(nums))
```
