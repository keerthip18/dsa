## 1. Creating lists

```python
empty = []
nums = [1, 2, 3, 4, 5]
mixed = [1, "two", 3.0, True, None, [6, 7]]   # any types, even other lists
from_string = list("hello")                   # ['h', 'e', 'l', 'l', 'o']
from_range = list(range(1, 6))                # [1, 2, 3, 4, 5]
zeros = [0] * 5                               # [0, 0, 0, 0, 0]
squares = [x * x for x in range(1, 6)]        # comprehension

print(empty, nums, mixed, from_string, from_range, zeros, squares, sep="\n")
```

## 2. Accessing items: indexing and slicing

```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
#  index:     0         1         2        3         4
#  negative: -5        -4        -3       -2        -1

print(fruits[0])     # apple  - first
print(fruits[-1])    # elderberry - last
print(fruits[1:3])   # ['banana', 'cherry']  - start included, end excluded
print(fruits[:2])    # first two
print(fruits[-2:])   # last two
print(fruits[::2])   # every second item
print(fruits[::-1])  # reversed copy
```

```python
# lists are mutable: change items in place
fruits = ["apple", "banana", "cherry"]
fruits[1] = "blueberry"
print(fruits)

# replace a slice (can even change the length)
fruits[1:3] = ["kiwi", "lemon", "mango"]
print(fruits)

# indexing past the end raises IndexError (slicing never does)
try:
    fruits[10]
except IndexError as e:
    print("IndexError:", e)
print(fruits[10:20])   # []
```

## 3. All list methods

`dir(list)` shows them. There are 11 (ignoring the `__dunder__` ones).

| Method | What it does | Returns |
|---|---|---|
| `append(x)` | add one item at the end | `None` |
| `extend(iterable)` | add every item from an iterable at the end | `None` |
| `insert(i, x)` | add an item at index `i` | `None` |
| `remove(x)` | delete the first item equal to `x` | `None` |
| `pop(i=-1)` | delete and **return** the item at `i` (last by default) | the item |
| `clear()` | delete everything | `None` |
| `index(x)` | position of the first `x` | `int` |
| `count(x)` | how many times `x` appears | `int` |
| `sort()` | sort in place | `None` |
| `reverse()` | reverse in place | `None` |
| `copy()` | shallow copy | new list |

```python
print([m for m in dir(list) if not m.startswith("_")])
```

### Adding items: `append`, `extend`, `insert`

```python
cart = ["milk"]
cart.append("bread")              # one item
print(cart)

cart.extend(["eggs", "butter"])   # many items
print(cart)

cart.insert(0, "coffee")          # at a position
print(cart)
```

```python
# append vs extend: append adds the list as ONE item
a = [1, 2]
a.append([3, 4])
print(a)          # [1, 2, [3, 4]]

b = [1, 2]
b.extend([3, 4])
print(b)          # [1, 2, 3, 4]

# extend works with any iterable, including strings
c = []
c.extend("hi")
print(c)          # ['h', 'i']
```

### Removing items: `remove`, `pop`, `clear` (and `del`)

```python
colors = ["red", "green", "blue", "green"]

colors.remove("green")     # removes the FIRST match only
print(colors)              # ['red', 'blue', 'green']

last = colors.pop()        # removes and returns the last item
print(last, colors)

first = colors.pop(0)      # removes and returns the item at index 0
print(first, colors)

colors.clear()
print(colors)              # []
```

```python
# errors to watch for
try:
    [1, 2].remove(99)
except ValueError as e:
    print("remove:", e)

try:
    [].pop()
except IndexError as e:
    print("pop:", e)

# del is a statement, not a method: delete by index or slice
nums = [0, 1, 2, 3, 4, 5]
del nums[0]
del nums[1:3]
print(nums)   # [1, 4, 5]
```

### Searching: `index`, `count`

```python
letters = ["a", "b", "c", "b", "a", "b"]

print(letters.count("b"))      # 3
print(letters.index("b"))      # 1  - first match
print(letters.index("b", 2))   # 3  - search starting at index 2

# index raises ValueError if missing, so check with `in` first
if "z" in letters:
    print(letters.index("z"))
else:
    print("z not found")
```

### Ordering: `sort`, `reverse`

```python
scores = [70, 95, 82, 61, 88]

scores.sort()
print(scores)                 # ascending

scores.sort(reverse=True)
print(scores)                 # descending

scores.reverse()              # just flips the current order, no sorting
print(scores)
```

```python
# sort with a key function: sort by something other than the item itself
words = ["banana", "Cherry", "apple", "fig"]

words.sort()
print(words)                  # uppercase comes before lowercase

words.sort(key=str.lower)
print(words)                  # case-insensitive

words.sort(key=len)
print(words)                  # by length
```

```python
# sort() changes the list and returns None; sorted() returns a NEW list
nums = [3, 1, 2]
result = nums.sort()
print(result)           # None  - a common mistake!

nums = [3, 1, 2]
new = sorted(nums)
print(nums, new)        # original untouched
```

### Copying: `copy`

```python
original = [1, 2, 3]
alias = original           # NOT a copy - same list, two names
duplicate = original.copy()

original.append(4)
print(alias)               # [1, 2, 3, 4] - changed too
print(duplicate)           # [1, 2, 3]   - independent
print(alias is original, duplicate is original)
```

## 4. Built-in functions and operators

```python
nums = [4, 8, 15, 16, 23, 42]

print(len(nums), min(nums), max(nums), sum(nums))
print(sum(nums) / len(nums))         # average
print(15 in nums, 99 not in nums)    # membership
print(any(n > 40 for n in nums))     # is at least one True?
print(all(n > 0 for n in nums))      # are all True?
print(sorted(nums, reverse=True))
print(list(reversed(nums)))
print([1, 2] + [3, 4])               # concatenation -> new list
print(["-"] * 3)                     # repetition
print([1, 2, 3] == [1, 2, 3], [1, 2] < [1, 3])   # compared item by item
```

```python
# enumerate, zip, map, filter all work on lists
names = ["John", "Amy", "Raj"]
for i, name in enumerate(names, start=1):
    print(i, name)

print(list(zip(names, [25, 30, 28])))
print(list(map(str.upper, names)))
print(list(filter(lambda n: len(n) > 3, names)))
```

## 5. Copying, and the shared-list trap

```python
# .copy(), list(x) and x[:] make SHALLOW copies: the outer list is new,
# but lists inside it are still shared
matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
shallow[0].append(99)
print(matrix)            # [[1, 2, 99], [3, 4]] - inner list changed too!

import copy
matrix = [[1, 2], [3, 4]]
deep = copy.deepcopy(matrix)
deep[0].append(99)
print(matrix)            # [[1, 2], [3, 4]] - untouched
```

```python
# the [[0] * 3] * 3 trap: three references to the SAME inner list
grid = [[0] * 3] * 3
grid[0][0] = 1
print(grid)              # [[1, 0, 0], [1, 0, 0], [1, 0, 0]]

# correct: a comprehension makes a new inner list each time
grid = [[0] * 3 for _ in range(3)]
grid[0][0] = 1
print(grid)              # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
```

```python
# don't remove items from a list while looping over it - items get skipped
nums = [1, 2, 2, 3, 4]
for n in nums:
    if n % 2 == 0:
        nums.remove(n)
print(nums)              # [1, 2, 3] - a 2 was skipped!

# build a new list instead
nums = [1, 2, 2, 3, 4]
nums = [n for n in nums if n % 2 != 0]
print(nums)              # [1, 3]
```

## 6. Lists as stacks and queues

- **Stack** (last in, first out): `append` + `pop()`. Both are fast.
- **Queue** (first in, first out): `pop(0)` is slow on big lists, because every other item has to shift. Use `collections.deque` instead.

```python
# stack: browser back button
history = []
history.append("home")
history.append("products")
history.append("cart")
print("back to:", history.pop())   # cart
print("back to:", history.pop())   # products
print(history)
```

```python
# queue: print jobs in arrival order
from collections import deque

jobs = deque(["report.pdf", "photo.png"])
jobs.append("invoice.pdf")         # join the back
while jobs:
    print("printing", jobs.popleft())   # serve the front
```

## 7. Real-world programs

### Shopping cart

```python
cart = []

def add_item(name, price, qty=1):
    cart.append({"name": name, "price": price, "qty": qty})

def remove_item(name):
    for item in cart:
        if item["name"] == name:
            cart.remove(item)
            return          # stop right after removing - safe
    print(f"{name} not in cart")

add_item("Notebook", 3.50, 4)
add_item("Pen", 1.20, 10)
add_item("Backpack", 45.00)
remove_item("Pen")

for item in cart:
    print(f"{item['name']:<10} {item['qty']:>2} x ${item['price']:>6.2f} = ${item['qty'] * item['price']:>7.2f}")
total = sum(item["qty"] * item["price"] for item in cart)
print(f"{'Total':<26} ${total:>7.2f}")
```

### To-do list

```python
todos = ["buy groceries", "pay bills", "call mom"]

todos.insert(0, "fix bug")       # urgent: put it first
todos.remove("pay bills")        # done
todos.append("book dentist")

for i, task in enumerate(todos, start=1):
    print(f"{i}. {task}")
```

### Leaderboard: top 3 scores

```python
players = [("Amy", 820), ("John", 950), ("Raj", 710), ("Lee", 990), ("Sam", 880)]

top3 = sorted(players, key=lambda p: p[1], reverse=True)[:3]
for rank, (name, score) in enumerate(top3, start=1):
    print(f"#{rank} {name:<5} {score}")
```

### Student grades report

```python
grades = [88, 92, 79, 65, 95, 70, 84]

average = sum(grades) / len(grades)
passed = [g for g in grades if g >= 70]
above_avg = [g for g in grades if g > average]

print(f"average:   {average:.1f}")
print(f"highest:   {max(grades)}, lowest: {min(grades)}")
print(f"passed:    {len(passed)}/{len(grades)}")
print(f"above avg: {above_avg}")
```

### Remove duplicates but keep the original order

```python
emails = ["a@x.com", "b@x.com", "a@x.com", "c@x.com", "b@x.com"]

# set() removes duplicates but loses the order
print(list(set(emails)))

# keep the first occurrence of each, in order
unique = []
for e in emails:
    if e not in unique:
        unique.append(e)
print(unique)

# shortcut: dict keys are unique and keep insertion order
print(list(dict.fromkeys(emails)))
```

### Pagination: split a list into pages

```python
products = [f"item{i}" for i in range(1, 12)]   # 11 items
page_size = 4

pages = [products[i:i + page_size] for i in range(0, len(products), page_size)]
for n, page in enumerate(pages, start=1):
    print(f"page {n}: {page}")
```

### Moving average of daily temperatures

```python
temps = [21, 23, 22, 25, 27, 26, 24]
window = 3

averages = []
for i in range(len(temps) - window + 1):
    chunk = temps[i:i + window]
    averages.append(round(sum(chunk) / window, 1))
print(averages)
```

### Undo / redo in a text editor

```python
text = ""
undo_stack, redo_stack = [], []

def type_text(new):
    global text
    undo_stack.append(text)
    redo_stack.clear()       # new typing wipes the redo history
    text += new

def undo():
    global text
    if undo_stack:
        redo_stack.append(text)
        text = undo_stack.pop()

def redo():
    global text
    if redo_stack:
        undo_stack.append(text)
        text = redo_stack.pop()

type_text("Hello")
type_text(" World")
print(repr(text))
undo(); print(repr(text))
redo(); print(repr(text))
```

### Inventory: sort, filter and restock

```python
inventory = [
    {"item": "laptop", "stock": 4, "price": 899},
    {"item": "mouse", "stock": 50, "price": 25},
    {"item": "monitor", "stock": 2, "price": 199},
    {"item": "keyboard", "stock": 15, "price": 49},
]

low_stock = [p["item"] for p in inventory if p["stock"] < 5]
print("restock:", low_stock)

by_price = sorted(inventory, key=lambda p: p["price"])
print("cheapest first:", [p["item"] for p in by_price])

total_value = sum(p["stock"] * p["price"] for p in inventory)
print(f"inventory value: ${total_value:,}")
```

### Tic-tac-toe board (2D list)

```python
board = [[" "] * 3 for _ in range(3)]
board[0][0] = "X"
board[1][1] = "O"
board[2][2] = "X"

for row in board:
    print(" | ".join(row))
    print("-" * 9)

# check a winning diagonal
diagonal = [board[i][i] for i in range(3)]
print("diagonal:", diagonal, "win" if diagonal.count("X") == 3 else "no win")
```
